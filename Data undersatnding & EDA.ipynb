# EDA

# 1. Import Libraries and Load Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\sp719\OneDrive\Desktop\project\ipl.csv")
```

# 2.Summary Statistics

```python
df.describe()#gives important statistical information
```

# 3.Missing Values

```python
df.isnull().sum()#missing values are present because some information is not applicable to every ball
```

# 4.Check Duplicates

```python
df.duplicated().sum()
```

# 5.Univariate Analysis

```python
#univariate analysis means studying one variable at a time
sns.histplot(df["runs_scored"], bins=10)
plt.title("Runs Distribution")
plt.show()
```

# 6.Bivariate Analysis

```python
#bivariate analysis means studying the relationship between two variables
sns.scatterplot(
    x="over",
    y="current_score",
    data=df.sample(5000)
)
plt.title("over vs current score")
plt.show()
```

# 7.Correlation Analysis

```python
#correlation tells us how strongly two numeric variables are related 
corr = df.select_dtypes("number").corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

# 8.Visualization

```python
#Toss decision-this compares how frequently teams chose bat and field after winning the toss
sns.countplot(x="toss_decision",data=df)
plt.title("Toss Decision")
plt.xlabel("Decision")
plt.ylabel("count")
plt.show()
```
