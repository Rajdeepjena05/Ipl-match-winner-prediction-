# Data Cleaning

# 1.Load the Dataset

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\sp719\OneDrive\Desktop\project\ipl.csv")
print(df.shape)
```

# 2.Handle Missing Values

```python
print(df.isnull().sum())#checking missing values
```

```python
#for columns where missing values have a meaningful category
df['city'] = df['city'].fillna('Unknown')
df['winner'] = df['winner'].fillna('Unknown')
df['player_of_match'] = df['player_of_match'].fillna('Unknown')
#for wicket-related columns missing means no wicket occurred so we can use 
df['wicket_kind'] = df['wicket_kind'].fillna('No wicket')
df['player_out'] = df['player_out'].fillna('No player')
df['fielder'] = df['fielder'].fillna('No fielder')
```

# 2.Remove Duplicates

```python
print(df.duplicated().sum())#checking duplicate rows
```

# 3.Fix incorrect Data types

```python
print(df.dtypes)
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df['result_margin'] = df['result_margin'].replace('undefined', pd.NA)
df['result_margin'] = pd.to_numeric(df['result_margin'])#now it becomes numeric column
```

# 4.Handle Outliers

```python
print(df['current_run_rate'].describe())
```

```python
Q1 = df['current_run_rate'].quantile(0.25)
Q3 = df['current_run_rate'].quantile(0.75)

IQR = Q3 - Q1

outliers = df[
    (df['current_run_rate'] < Q1 - 1.5 * IQR) |
    (df['current_run_rate'] > Q3 + 1.5 * IQR)
]

print(len(outliers))
```
